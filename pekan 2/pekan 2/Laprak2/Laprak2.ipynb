## 2.1 Dasar Teori
Pengolahan informasi di dalam program membutuhkan sarana penyimpanan dan pengelompokan jenis nilai. Variabel berperan sebagai wadah penampung atau referensi nilai di dalam memori komputer. Pembuatan variabel pada Python dilakukan dengan menetapkan nilai menggunakan tanda sama dengan (=). Sementara itu, tipe data berfungsi menetapkan sifat dan jenis nilai yang dimuat oleh variabel tersebut sehingga sistem dapat memprosesnya dengan tepat.

Python memiliki beberapa tipe data standar yang sering digunakan dalam penyusunan program, antara lain:

1. Integer (int): Digunakan untuk menampung data berupa angka atau bilangan bulat tanpa nilai pecah, seperti 10, 25, maupun bilangan negatif seperti -5.

2. Float (float): Digunakan untuk menyimpan nilai angka pecahan atau desimal, seperti 3.14, 10.5, dan 0.75.

3. Complex (complex): Tipe data khusus untuk merepresentasikan bilangan kompleks yang terdiri atas komponen real serta imajiner, contohnya 3+2j.

4. String (str): Digunakan untuk menampung data teks atau deretan karakter. Penulisan nilai string diapit oleh tanda petik, seperti "Python" atau "Informatika".

5. Boolean (bool): Tipe data logika yang hanya memiliki dua pilihan nilai kebenaran, yaitu True (benar) dan False (salah). Tipe ini umumnya dipakai untuk mengevaluasi suatu kondisi.

## 2.2 Kode Program

### 2.2.1 Program "Numerik_2611531010"

````python
#create a variable with numeric type value.
a=1010
print("The type of variable having value",a,"is",type(a))

#create a variable with float value.
b=10.2345
print("The type of variable having value",b,"is",type(b))

#create a variable with complex value.
c=100+3j
print("The type of variable having value",c,"is",type(c))
````

```text
The type of variable having value 1010 is <class 'int'>
The type of variable having value 10.2345 is <class 'float'>
The type of variable having value (100+3j) is <class 'complex'>


**Penjelasan :**

1. `a = 1010` digunakan untuk membuat variabel `a` yang bernilai angka bulat (*integer*).
2. `b = 10.2345` digunakan untuk membuat variabel `b` yang bernilai angka desimal (*float*).
3. `c = 100+3j` digunakan untuk membuat variabel `c` yang bernilai bilangan kompleks (*complex*).
4. `type()` digunakan untuk mengecek dan menampilkan tipe data dari suatu variabel.
5. `print()` digunakan untuk menampilkan teks, nilai variabel, serta tipe data variabel ke layar.

### 2.2.3 Program "Boolean_2611531010"

````python
#Deklarasi variabel dengan tipe data Boolean
is_lulus = True
is_cumlaude = True

# Menggunakan Boolean
nilai = 85
batas_lulus = 75

# Menentukan nilai Boolean dari kondisi 
status_kelulusan = nilai >= batas_lulus  #Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai)
print("Apakah Lulus?:", status_kelulusan)
if is_lulus and is_cumlaude:
    print("Selamat, Anda lulus dengan predikat cumlaude!")
````

```text
=== Check Kelulusan ===
Nilai: 85
Apakah Lulus?: True
Selamat, Anda lulus dengan predikat cumlaude!


**Penjelasan:**

1. Variabel `is_lulus` dan `is_cumlaude` menyimpan nilai bertipe data Boolean, yaitu `True`.
2. Variabel `status_kelulusan` menyimpan hasil dari kondisi `nilai >= batas_lulus`, yang menghasilkan nilai `True` karena angka 85 lebih besar dari 75.
3. Pengondisian `if` menggunakan operator logika `and` untuk memastikan kedua variabel (`is_lulus` dan `is_cumlaude`) bernilai `True` sebelum menjalankan baris kode di dalamnya.
4. Fungsi `print()` menampilkan teks judul, isi nilai variabel, hasil pengecekan Boolean, serta pesan selamat ke layar.

### 2.2.4 Program "Input_2611531010"

````python
# Nama program untuk menghitung luas persegi panjang
# Nama variabel ditambah 4 digit nim terakhir contoh: panjang_1010
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nila panjang dann lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_1234 = input('Masukkan nilai panjang')
lebar_1234 = input('Masukkan nilai lebar')
luas_1234 = int(panjang_1234) * int(lebar_1234)
print("Luas =", luas_1234)
````

```text
Masukkan nilai panjang 10
Masukkan nilai lebar 5
Luas = 50

**Penjelasan:**

1. Fungsi `input()` digunakan untuk menerima masukan nilai panjang dan lebar dari pengguna secara interaktif melalui layar terminal.
2. Fungsi `int()` mengonversi tipe data masukan yang semula berbentuk teks (*string*) menjadi bilangan bulat (*integer*) agar dapat dihitung secara matematis.
3. Perkalian antara variabel `panjang_1234` dan `lebar_1234` dilakukan untuk menghitung nilai luas persegi panjang yang kemudian disimpan ke dalam variabel `luas_1234`.
4. Fungsi `print()` menampilkan hasil perhitungan nilai luas ke layar.

### 2.2.5 Program "Konstanta_2611531010"

````python
# Program ini menggunakan konstanta untul menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1010

from typing import Final
PI: Final = 3.14
print("pi: %f" %(PI))
jari_1234 = float(input('Masukkan nilai jari-jari: '))
luas_1234 = PI * jari_1234 * jari_1234
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" %(jari_1234, luas_1234))
````

```text
pi: 3.140000
Masukkan nilai jari-jari: 7
Luas lingkaran dengan jari-jari 7.00 adalah 153.86

**Penjelasan:**

1. Modul `typing.Final` digunakan untuk menandai variabel `PI` sebagai nilai konstanta yang bernilai tetap, yaitu `3.14`.
2. Fungsi `input()` yang dibungkus `float()` mengambil masukan jari-jari dari pengguna dan langsung mengonversinya menjadi bilangan desimal.
3. Rumus `PI * jari_1234 * jari_1234` digunakan untuk menghitung nilai luas lingkaran yang kemudian disimpan ke variabel `luas_1234`.
4. Fungsi `print()` menggunakan format penulisan `%f` untuk menampilkan nilai desimal, serta `%.2f` untuk membatasi tampilan angka di belakang koma menjadi dua digit.

Berdasarkan praktikum yang telah dilaksanakan, dapat ditarik beberapa kesimpulan sebagai berikut:

1. Variabel dan Tipe Data: Variabel berfungsi sebagai tempat penyimpanan nilai di dalam memori komputer yang dibuat menggunakan operator penugasan (=). Tipe data menentukan jenis nilai yang disimpan, meliputi int (bilangan bulat), float (bilangan desimal), complex (bilangan kompleks), str (teks/karakter), dan bool (nilai kebenaran True/False).

2. Konstanta: Dalam Python, konstanta diwakili dengan penamaan variabel menggunakan huruf kapital (seperti PI) atau penanda Final untuk menandakan bahwa nilainya bersifat tetap dan tidak dimaksudkan untuk diubah selama program berjalan.

3. Input, Output, dan Konversi Data: Fungsi input() digunakan untuk menerima masukan pengguna yang secara otomatis bertipe str, sehingga memerlukan konversi tipe data (type casting) seperti int() atau float() jika ingin digunakan dalam perhitungan matematis. Sementara itu, fungsi print() berfungsi untuk menampilkan hasil proses atau nilai variabel ke layar terminal.
