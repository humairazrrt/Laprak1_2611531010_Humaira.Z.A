# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori
Praktikum ini membahas pengenalan bahasa pemrograman Python, penggunaan Visual Studio Code sebagai alat untuk membuat program, serta pemanfaatan GitHub dalam menyimpan dan mengelola kode.

1. Python
Python adalah bahasa pemrograman tingkat tinggi dengan sintaks yang sederhana dan mudah dipelajari. Python dapat digunakan dalam berbagai bidang, seperti pembuatan aplikasi, pengolahan data, dan pengembangan perangkat lunak.
2. Visual Studio Code
Visual Studio Code (VS Code) merupakan aplikasi editor kode yang digunakan untuk menulis, mengedit, dan menjalankan program. VS Code mendukung banyak bahasa pemrograman, termasuk Python, serta memiliki fitur seperti terminal, debugging, dan integrasi Git.
3. Git
Git adalah sistem pengontrol versi yang berfungsi untuk mencatat dan mengelola perubahan dalam suatu proyek. Penggunaan Git memudahkan programmer dalam mengatur berbagai versi kode program.
4. GitHub
GitHub adalah platform daring yang digunakan untuk menyimpan repository berbasis Git. GitHub memungkinkan kode program dicadangkan secara online, diakses dari berbagai perangkat, serta digunakan untuk berkolaborasi dengan pengguna lain.
5. Repository
Repository adalah tempat yang digunakan untuk menyimpan sebuah proyek beserta kode dan file pendukungnya. Repository dapat berada di komputer secara lokal maupun tersimpan secara online di GitHub.
6. Commit dan Push
Commit merupakan proses mencatat perubahan yang telah dibuat pada repository lokal. Sementara itu, push digunakan untuk mengirim perubahan tersebut dari repository lokal ke repository yang tersimpan di GitHub.
